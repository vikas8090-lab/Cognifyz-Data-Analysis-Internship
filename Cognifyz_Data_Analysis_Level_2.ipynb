# Load the Dataset

```python
import pandas as pd

file_path = '/Dataset .csv(cognifyz).csv'

df = pd.read_csv(file_path, encoding='latin1')

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

# Display the Dataset Columns

```python
print("Column Names:")
print(df.columns.tolist())
```

# Level 2 – Task 1

# Restaurant Ratings
* Analyze the Distribution of Aggregate Ratings

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))

plt.hist(df['Aggregate rating'], bins=10)

plt.title('Distribution of Aggregate Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Number of Restaurants')

plt.show()
```

* Identify the Most Common Rating Range

```python
rating_ranges = pd.cut(
    df['Aggregate rating'],
    bins=[0, 1, 2, 3, 4, 5],
    include_lowest=True
)

most_common_range = rating_ranges.value_counts().idxmax()

print("Most Common Rating Range:", most_common_range)
```

* Calculate the Average Number of Votes

```python
average_votes = df['Votes'].mean()

print("Average Number of Votes:", round(average_votes, 2))
```

# Level 2 – Task 2

# Cuisine Combination
* Identify the Most Common Cuisine Combinations

```python
cuisine_combinations = df['Cuisines'].value_counts()

print("Top 10 Most Common Cuisine Combinations:")
print(cuisine_combinations.head(10))
```

* Analyze Ratings of Cuisine Combinations

```python
cuisine_rating = df.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False)

print("Top Cuisine Combinations by Average Rating:")
print(cuisine_rating.head(10))
```

# Level 2 – Task 3


# Geographic Analysis
* Plot Restaurant Locations

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df['Longitude'],
    df['Latitude'],
    alpha=0.5
)

plt.title('Geographic Distribution of Restaurants')
plt.xlabel('Longitude')
plt.ylabel('Latitude')

plt.show()
```

# Identify Geographic Clusters and Patterns

### Geographic Clusters and Patterns

- The restaurants are distributed across multiple geographic locations.
- A dense concentration of restaurants can be observed around longitude 70–100 and latitude 10–30.
- Another noticeable concentration is visible around longitude -120 to -90 and latitude 30–50.
- A smaller concentration can also be observed around longitude 140–160 and latitude -40 to -30.
- Several restaurants are scattered across other geographic locations.

# Level 2 – Task 4

# Restaurant Chains
* Identify Restaurant Chains

```python
restaurant_counts = df['Restaurant Name'].value_counts()

print("Top 20 Restaurant Chains:")
print(restaurant_counts.head(20))
```

* Analyze Ratings of Restaurant Chains

```python
chain_ratings = df.groupby('Restaurant Name')['Aggregate rating'].mean().sort_values(ascending=False)

print("Top 20 Restaurant Chains by Average Rating:")
print(chain_ratings.head(20))
```

* Analyze Popularity of Restaurant Chains

```python
chain_popularity = df['Restaurant Name'].value_counts()

print("Top 20 Most Popular Restaurant Chains:")
print(chain_popularity.head(20))
```
